A variable is one of the basic notions in programming. Every variable has a value such as a
number or a piece of text. Every value has a type, and the type determines which operations
are possible.

Basic types hold a single value. Data structures hold several of them and provide a way of
reaching each one.

## Values and types

Python has four basic types for single values.

```python
print(type(31000))
print(type(4.5))
print(type("Neuchâtel"))
print(type(True))
```

`int` is whole numbers, `float` numbers with a decimal part, `str` text, and `bool` either
`True` or `False`.

Substraction (`-`) is defined for numbers. Applied to text it raises a `TypeError`.

```python
print(31000 - 4500)
print("31000" - "4500")
```

`+` is defined for both types, and it does something different in each case. On numbers it
adds. On text it joins the two pieces end to end, which is called concatenation.

```python
print(38000 + 41000)
print("Neuchâtel" + " " + "2026")
```

Both results are correct for the types given, and concatenation is often what is wanted. The
difficulty arises when a figure has arrived as text, because `+` then joins it instead of adding
it, and no error is raised.

```python
q1 = "38000"
q2 = "41000"

print(q1 + q2)
```

The two quarters sum to 79000. The line above printed the two figures written one after the
other.

### None

Similar to `null` in databases and other programming languages, `None` means no value. It is
neither zero nor empty text.

```python
amount = None

print(amount)
print(type(amount))
print(amount is None)
```

```python
amount = None

print(amount + 100)
```

`None` has to be detected before it reaches a calculation. The test is `is None`, written
with `is` rather than `==`.

```python
readings = [38000, None, 39500, None, 47000]

total = 0
counted = 0

for item in readings:
    if item is not None:
        total = total + item
        counted = counted + 1

print("Total:", total, "from", counted, "of", len(readings), "readings")
```

### Converting

Sometimes, when using data from external sources such as a CSV export or a web form, figures
arrive as text and need to be converted before being used in arithmetic. For that we can use the
functions `int()`, `float()` and `str()`.

```python
print(int("4500") + 500)
print(float("4.5") * 2)
print("Total: " + str(31000))
```

Conversion fails when the text does not represent a number.

```python
print(int("n/a"))
```

`int()` applied to a number with decimals does not round it, it cuts it towards zero. `round` is
the function for rounding.

```python
print(int(4500.75))
print(int(-4500.75))
print(round(4500.75))
```

Be careful when converting to boolean values, `bool()` applied to text returns `True` for any text that is not empty, including `"False"` and
`"0"`.

```python
print(bool("False"))
print(bool("0"))
print(bool(""))
```

## How Python checks types

Python is **dynamically typed**. Put in plain terms: the type belongs to the value, not to the
name, so assigning a new value to the same name changes the type as well. The type is checked
while the program runs, at the moment an operation is performed on the value.

```python
x = 5
print(x, type(x))

x = "five"
print(x, type(x))

x = True
print(x, type(x))
```

This is dynamic typing. In a **statically typed** language such as Java or C#, you write
`int x = 5;`, which declares that `x` holds whole numbers, and assigning text to it later is
rejected before the program is allowed to run.

The difference is when the check happens. A statically typed language checks before execution
starts. Python checks when it performs an operation on the value.

```python
readings = [38000, 41000, 39500, "44000", 47000]

total = 0
for item in readings:
    total = total + item
    print("added", item, "running total", total)

print("Total:", total)
```

The loop added three values before reaching the text and stopping. Python did not report the
problem when the list was created, but when the addition was attempted. Dynamic typing spares the
programmer the work of declaring types, and this is one of the reasons why programming in Python
is popular, but in exchange type errors may go undetected and surface only during execution.

## Lists

A value that holds other values is a **data structure**. What distinguishes one from another is
whether the order of items is kept, whether an item is reached by position or by name, and whether
it can be changed after it is created. Python has four built-in data structures: lists, tuples,
dictionaries and sets. This notebook covers lists and dictionaries.

A list holds values in order. It is written with square brackets.

```python
revenue = [38000, 41000, 39500, 44000, 47000, 45500]

print(revenue)
print(len(revenue))
```

An item is read by its position. Positions start at 0, and negative positions count backwards
from the end.

```python
print(revenue[0])
print(revenue[1])
print(revenue[-1])
```

```python
print(revenue[6])
```

The last item of a six-item list is at position 5, so position 6 does not exist. `IndexError`
usually means a loop ran one step too far.

A **slice** returns part of a list. It is written as two positions separated by a colon. The
first position is included, the second is not.

```python
print(revenue[0:3])
print(revenue[3:])
```

### Changing a list

```python
revenue[0] = 38500
print(revenue)

revenue.append(49000)
print(revenue)
```

`len(revenue)` is a function, and the list is its argument. `append` is a **method**, a function
that belongs to the list and is called on it, written `object.method(arguments)`. Notebook 9
covers the distinction.

### Looping

The `for` loop works directly on a list, so neither `range` nor the positions are needed.

```python
total = 0

for amount in revenue:
    total = total + amount

print("Total:", total)
```

This is the accumulator pattern from notebook 4, with the values coming from a list.

When the position is needed as well as the value, loop over the positions.

```python
for i in range(len(revenue)):
    print("month", i + 1, ":", revenue[i])
```

Python provides built-in functions for operations that are needed often.

```python
print("total  ", sum(revenue))
print("highest", max(revenue))
print("lowest ", min(revenue))
print("average", round(sum(revenue) / len(revenue), 2))

print(sorted(revenue))
print(revenue)
```

`sorted` returns a new list and leaves the original unchanged.

### Two names for one list

Because a list can be changed, assigning it to a second name needs care. Run the cell and compare
the two lists at the end.

```python
budget = [12000, 8000, 5000]
scenario = budget

scenario[0] = 20000

print("scenario:", scenario)
print("budget:  ", budget)
```

`scenario = budget` did not create a second list. It attached a second name to the existing
one, so a change through either name is visible through both.

```python
a = 10
b = a
b = 20

print("a:", a)
print("b:", b)
```

Here `a` is unchanged, because `b = 20` replaces what the name refers to.
`scenario[0] = 20000` reaches into the list and modifies it.

Lists are **mutable**, so their contents can change in place. Numbers, text and booleans are
**immutable**, so operations on them produce a new value.

```python
budget = [12000, 8000, 5000]
scenario = budget.copy()

scenario[0] = 20000

print("scenario:", scenario)
print("budget:  ", budget)
```

`copy` creates a second list with the same contents. Without it, a change made through one name
also changes the list seen through the other.

## Dictionaries

In a list, a value is reached by its position. That is suitable when the position itself has a
meaning, such as the month number in a list of twelve monthly figures. It is unsuitable when the
values are identified by a name.

Below, two lists hold a cost breakdown. Finding the rent means finding its position first.

```python
categories = ["salaries", "rent", "software", "travel", "other"]
amounts = [31000, 4200, 2800, 1500, 2500]

i = categories.index("rent")
print(amounts[i])
```

The two lists also have to stay the same length and in the same order. Adding a category to
one and forgetting the other breaks the lookup.

```python
categories.append("marketing")

print(len(categories), len(amounts))
print(amounts[categories.index("marketing")])
```

A dictionary stores each value together with the name that identifies it. The name is called the
**key**, and the value is reached directly by its key, so neither problem arises. A dictionary is
written with curly brackets, in `key: value` pairs.

```python
costs = {
    "salaries": 31000,
    "rent": 4200,
    "software": 2800,
    "travel": 1500,
    "other": 2500,
}

print(costs["rent"])
print(len(costs))
```

```python
print(costs["marketing"])
```

A missing key raises `KeyError`. `get` returns a default instead of stopping.

```python
print(costs.get("marketing", 0))
```

### Adding and changing

Assigning to an existing key replaces its value. Assigning to a new key adds it. Both use the same
syntax, so a mistyped key adds an entry rather than reporting an error.

```python
costs["rent"] = 4500
costs["marketing"] = 3000

print(costs)
```

`in` reports whether a key is present and returns `True` or `False`.

```python
print("rent" in costs)
print("bonuses" in costs)

if "travel" in costs:
    print("Travel budget:", costs["travel"])
```

### Looping

Looping over a dictionary gives the keys. `items()` gives the key and the value together, which
avoids a second lookup.

```python
for category in costs:
    print(category, ":", costs[category])
```

```python
total = 0

for category, amount in costs.items():
    total = total + amount

print("Total costs:", total)
```

Two loop variables follow `for` because each item is a pair.

## When the choice matters

The cost breakdown is small, so either structure would work. The difference shows on larger data,
and it comes from how each one finds a value.

A list has to be searched. Python compares the value against the items one at a time until it
finds a match or reaches the end, so the time grows with the length of the list.

A dictionary is not searched. It computes the storage location from the key, so a lookup takes
about the same time whatever the size.

### Reconciling payments against invoices

A company issued 12000 invoices and received payments referencing some of them. The question is
which invoices remain unpaid.

```python
import random

rnd = random.Random(3)

invoices = []
for i in range(12000):
    invoices.append("INV-" + str(10000 + i))

payments = rnd.sample(invoices, 9600)
rnd.shuffle(payments)

print("invoices:", len(invoices))
print("payments:", len(payments))
```

Searching the payments for each invoice puts one loop inside another. This cell takes a few
seconds.

```python
import time

start = time.time()

unpaid = 0
for invoice in invoices:
    found = False
    for payment in payments:
        if payment == invoice:
            found = True
            break
    if not found:
        unpaid = unpaid + 1

nested_time = time.time() - start

print("Unpaid invoices:", unpaid)
print("Time taken:", round(nested_time, 3), "seconds")
```

The same question with a dictionary. Each payment reference becomes a key.

```python
start = time.time()

paid = {}
for payment in payments:
    paid[payment] = True

unpaid = 0
for invoice in invoices:
    if invoice not in paid:
        unpaid = unpaid + 1

dict_time = time.time() - start

print("Unpaid invoices:", unpaid)
print("Time taken:", round(dict_time, 4), "seconds")
print("Faster by a factor of:", round(nested_time / dict_time))
```

Both versions report the same count, and the second is several hundred times faster. The first
version makes one comparison for every pair of invoice and payment. The second passes once through
each list.

```python
for size in [3000, 6000, 12000]:
    inv = []
    for i in range(size):
        inv.append("INV-" + str(10000 + i))
    pay = rnd.sample(inv, int(size * 0.8))

    start = time.time()
    for invoice in inv:
        for payment in pay:
            if payment == invoice:
                break
    elapsed = time.time() - start

    print(size, "invoices:", round(elapsed, 3), "seconds")
```

Doubling the invoices roughly quadruples the time, because both loops doubled and the work is
their product. At 120000 invoices the first version would take several minutes. The second would
still be under a second.

### 7. Reconciliation, with amounts

The invoice amounts are in a separate dictionary this time. The slow version is written for you.

**a)** Run it and note the time.

**b)** Write the dictionary version. It should give the same two figures: how many invoices are
unpaid, and the total value outstanding.

**c)** Print how many times faster your version is.

**d)** In a sentence: at 60000 invoices, roughly how much longer would each version take?

```python
import random, time

rnd = random.Random(11)

invoice_refs = []
invoice_amounts = {}
for i in range(10000):
    ref = "F-" + str(20000 + i)
    invoice_refs.append(ref)
    invoice_amounts[ref] = rnd.randrange(500, 9000)

payment_refs = rnd.sample(invoice_refs, 8200)
rnd.shuffle(payment_refs)

print("invoices:", len(invoice_refs))
print("payments:", len(payment_refs))
```

```python
start = time.time()

unpaid_count = 0
unpaid_value = 0

for ref in invoice_refs:
    found = False
    for payment in payment_refs:
        if payment == ref:
            found = True
            break
    if not found:
        unpaid_count = unpaid_count + 1
        unpaid_value = unpaid_value + invoice_amounts[ref]

slow_time = time.time() - start

print("Unpaid invoices:", unpaid_count)
print("Outstanding value:", unpaid_value)
print("Time taken:", round(slow_time, 3), "seconds")
```

### 8. A report with two faults

The program finds which clients account for more than 10% of total revenue. Each position in the
two lists is one invoice, and a client can appear more than once.

It runs cleanly, and both the total and the list of clients are wrong. One fault is in how revenue
per client is built up, the other in how the threshold is calculated.

Find both, correct them, and say for each what effect it had. Then check each client in the
corrected list against the 10% threshold by hand.

```python
clients = ["Aurex", "Belmont", "Cadence", "Aurex", "Delta", "Belmont", "Echo", "Aurex"]
amounts = [42000, 18000, 31000, 27000, 9000, 15000, 54000, 12000]

revenue = {}
for i in range(len(clients)):
    revenue[clients[i]] = amounts[i]

running = 0
major = []
for name in revenue:
    running = running + revenue[name]
    if revenue[name] > running * 0.10:
        major.append(name)

print("Total revenue:", running)
print("Clients above 10%:", major)
```
