This notebook covers two topics: functions and errors.

A **function** is a named block of code that performs a task. Python provides many functions
ready to use, and it also lets you write your own. Writing a function means the code for a task is
written once and then used wherever it is needed, and the program becomes easier to read, because
the function's name says what the code does.

An **error** is what happens when a program cannot do what its code asks. Some errors stop the
program, and some go unnoticed. This notebook shows the different kinds of errors, how a program
can handle an error and continue, and how a function can report an error of its own when it
receives values it cannot work with.

Most topics are followed by an exercise on what they covered.

## Using ready-made functions

Several functions have already been used in this course: `print`, `len`, `round`, `sum`, `max`,
`int` and `float`. Using a function is called **calling** it. A call is written as the function's
name followed by parentheses. The values written between the parentheses are the **arguments** of
the call, and they are the input the function works on. Most functions then give back a result,
called the **return value**, which can be stored in a variable or used in a calculation.

````python
revenue = [38000, 41000, 39500, 44000, 47000, 95000]

count = len(revenue)
highest = max(revenue)
average = round(sum(revenue) / count, 2)

print(count, highest, average)
````

In the first line after the list, `len` is called with the argument `revenue`, and its return
value, `6`, is stored in `count`. The last calculation combines three calls: `sum` returns the
total, the total is divided by `count`, and `round` is called with two arguments, the value to
round and the number of decimals.

## Writing your own functions

Ready-made functions cover tasks that many programs need. For tasks that are specific to a given
piece of work, Python lets you define your own functions, which are then called in the same way as
the ready-made ones.

### Why write a function

The example below works with amounts that were exported from an accounting system as text. Some of
the amounts use a space or an apostrophe to separate the thousands, as in `"41 000"` or
`"1'250.00"`. Before such an amount can be used in a calculation, the space or apostrophe has to be
removed and the remaining text converted to a number.

The first cell computes the exact total of the amounts. The second computes the total after
rounding each amount to a whole number.

````python
export = ["38000", "41 000", "39500.50", "1'250.00", "2 750.25"]

total = 0
for value in export:
    cleaned = value.replace(" ", "").replace("'", "")
    total = total + float(cleaned)

print("Exact total:", total)
````

````python
total_rounded = 0
for value in export:
    cleaned = value.replace(" ", "").replace("'", "")
    total_rounded = total_rounded + round(float(cleaned))

print("Rounded total:", total_rounded)
````

Both loops contain the same cleaning line:

```python
cleaned = value.replace(" ", "").replace("'", "")
```

Repeating this line has two drawbacks. Every further calculation on the same data needs another
copy of it. And if the cleaning ever has to change, for example to remove commas as well, every
copy has to be found and corrected, and a copy that is missed will give a different result from
the others.

A function avoids both problems: the cleaning is written once, inside the function, and the
function is called wherever the cleaning is needed.

### Defining and calling a function

The cell below defines a function called `clean_amount`, which carries out the cleaning in three
steps, and then calls it twice.

````python
def clean_amount(value):
    without_spaces = value.replace(" ", "")
    without_separators = without_spaces.replace("'", "")
    amount = float(without_separators)
    return amount


print(clean_amount("41 000"))
print(clean_amount("1'250.00"))
````

The definition has the following parts:

- `def` is the keyword that starts a function definition.
- `clean_amount` is the name of the function.
- `value`, between the parentheses, is a **parameter**. A parameter is a name that stands for the
  argument the function receives each time it is called.
- The colon ends the first line, and the indented lines below it form the **body** of the function,
  in the same way as the body of a loop or of an `if`.
- The body runs from top to bottom: it removes the spaces, then the apostrophes, then converts the
  result to a number.
- `return amount` ends the function and sends the value of `amount` back to the place where the
  function was called. That value is the return value of the call.

Defining a function does not run it. The body runs only when the function is called. In the call
`clean_amount("41 000")`, the argument is `"41 000"`, and during that call the parameter `value`
refers to it.

The two totals from above can now be computed with the function:

````python
total = 0
total_rounded = 0

for value in export:
    total = total + clean_amount(value)
    total_rounded = total_rounded + round(clean_amount(value))

print("Exact total:", total)
print("Rounded total:", total_rounded)
````

The results are the same as before. The cleaning is now written in one place only, and the loop
reads as a description of what is computed: clean each amount and add it to the total.

A function can also have more than one parameter. The function below has two, a list of amounts
and a threshold, and it counts how many of the amounts exceed the threshold, calling
`clean_amount` for each of them.

````python
def count_above(values, threshold):
    count = 0
    for value in values:
        if clean_amount(value) > threshold:
            count = count + 1
    return count


print(count_above(export, 10000))
print(count_above(export, 39000))
````

When a function has several parameters, the arguments are matched to them in order: in the first
call, `values` refers to `export` and `threshold` to `10000`. The `return` is placed after the
loop, not inside it, so that the count is returned only once all the amounts have been checked.

### Documenting a function

A function can carry a short description of what it does, written as text in triple quotes on the
first line of its body. This text is called a **docstring**. It states what the function expects
and what it returns, and the built-in function `help` displays it.

````python
def clean_amount(value):
    """Convert an amount stored as text into a number, ignoring spaces and apostrophes."""
    without_spaces = value.replace(" ", "")
    without_separators = without_spaces.replace("'", "")
    amount = float(without_separators)
    return amount


help(clean_amount)
````

## Returning values and combining calls

### Returning a value

`return` ends the function and sends a value back to the code that called it. A common mistake is
to print the result inside the function instead of returning it. The function below computes the
Swiss VAT at the standard rate of 8.1% and prints it.

````python
def calculate_vat(amount):
    vat = amount * 0.081
    print(round(vat, 2))
    print("End of function call")
    print("--")


result = calculate_vat(1000)
print("VAT on total amount as calculated:", result)
````

The VAT, `81.0`, appears on the screen, because the function prints it. But the variable
`result` holds `None`. `None` is Python's value for "no value", and it is what every function
returns when its body contains no `return` statement.

Because the function returns `None`, its result cannot be used in a calculation:

````python
print(1000 + calculate_vat(1000))
````

Python cannot add a number and `None`, so it raises a `TypeError`.

The version below returns the VAT instead of printing it, and the price including VAT can then be
computed from it:

````python
def vat_on(amount):
    vat = amount * 0.081
    return round(vat, 2)


price = 1000
print(price + vat_on(price))
````

`print` and `return` therefore do different things. `print` displays a value on the screen for
a person to read. `return` hands a value to the rest of the program so that it can be used. A
function whose result is needed later must return it.

### Combining function calls

Because a call to a function that returns a value produces that value, the call can be written
anywhere the value itself could be written, including as the argument of another call. Writing one
call inside another is called **composition**.

The two cells below compute the VAT on an amount stored as text. The first does it in two steps,
storing the intermediate result in a variable. The second writes the call to `clean_amount`
directly as the argument of `vat_on`.

````python
amount_text = "1'250.00"

amount = clean_amount(amount_text)
tax = vat_on(amount)

print(tax)
````

````python
print(vat_on(clean_amount(amount_text)))
````

Both give the same result. In the composed version, Python works from the inside out: it first
calls `clean_amount("1'250.00")`, which returns `1250.0`, then calls `vat_on(1250.0)`, which returns
`101.25`, and finally passes that value to `print`.

The first section of this notebook already contained a composition: in
`round(sum(revenue) / count, 2)`, the value returned by `sum` is used in a division, and the result
of the division is the first argument of `round`.

Composition only works with functions that return their result. The first version of `calculate_vat` could not be used in
this way, since it returns `None`. When a composition becomes long, splitting it into steps with
named variables, as in the first cell, makes it easier to read and to check.

## Variables and arguments

### Local variables

A variable that is created inside a function exists only while the function is running. Such a
variable is called a **local variable**. Once the function has finished, the variable no longer
exists, and code outside the function cannot use it.

````python
def vat_on(amount):
    vat = amount * 0.081
    return round(vat, 2)

print(vat_on(1000))
print(vat)
````

The call works and prints `81.0`. The last line fails with a `NameError`, because `vat` was
created inside `vat_on` and does not exist outside it. Parameters are local variables as well: a
parameter exists only for the duration of the call.

### Default values and keyword arguments

A parameter can be given a **default value**, which is used when a call does not supply an argument
for that parameter. In the function below, the discount rate is 10% unless another rate is given.

````python
def apply_discount(amount, rate=0.1):
    discounted = amount * (1 - rate)
    return round(discounted, 2)


print(apply_discount(1000))
print(apply_discount(1000, 0.25))
````

The first call gives no rate, so the default of `0.1` is used and the result is `900.0`. The
second call supplies `0.25`, and the result is `750.0`.

An argument can also be passed by naming the parameter it is meant for, as in `rate=0.25`. This is
called a **keyword argument**. It makes the call easier to read, because it is clear which value is
which, and the arguments can then be given in any order.

````python
print(apply_discount(1000, rate=0.25))
print(apply_discount(rate=0.25, amount=1000))
````

## Recursion

A function can call itself. This is called **recursion**. For recursion to work, two conditions
must hold. Each call must work on a smaller version of the problem than the call before it. And
there must be a case, called the **base case**, in which the function returns a result without
calling itself again.

The function below computes the balance of an investment after a number of years, with interest
added once a year.

````python
def balance_after(capital, rate, years):
    if years == 0:
        return capital
    return balance_after(capital * (1 + rate), rate, years - 1)


print(round(balance_after(10000, 0.03, 5), 2))
````

The call `balance_after(10000, 0.03, 5)` adds one year of interest and calls
`balance_after(10300.0, 0.03, 4)`. That call adds another year and calls the function with 3 years
left, and so on. When `years` reaches 0, the base case applies: the function returns the capital
without calling itself, and that value is passed back through all the earlier calls.

If the base case were missing, or were never reached, the function would keep calling itself
without end, in the same way that a `while` loop without a stop condition keeps running. Python
prevents this by stopping the program with a `RecursionError` after about 1000 nested calls.

The same result can be computed with a loop, and for a calculation like this one a loop is usually
easier to read. Recursion is the easier choice when **data is nested and the depth isn’t known in advance**. A loop would need to know beforehand how many levels to go down, while a recursive function handles each level the same way, however many there are.

## Errors

### Three kinds of errors

A program can go wrong in three different ways. The three kinds differ in when the mistake shows
up, and in whether Python notices it at all.

1. A **syntax error** means that the code is not valid Python. Python notices it before running
   the code.
2. A **runtime error**, also called an **exception**, means that the code is valid Python, but an
   operation fails while the program is running. Python notices it at the line where it happens
   and stops the program there.
3. A **logic error** means that the program runs to the end without any error message, but its
   result is wrong. Python does not notice it.

The three cells below show one example of each.

**A syntax error.** In the cell below, the colon at the end of the `if` line is missing.

````python
total = 1500

print("checking the total")
if total > 1000
    print("large")
````

The message `checking the total` is not printed, because Python checks the whole cell before
running any of it, and stopped at the missing colon. A syntax error can only be removed by
correcting the code.

**A runtime error.** In the cell below, the code is valid, but the last value in the list cannot be
converted to a number.

````python
figures = ["38000", "41000", "n/a"]

for value in figures:
    print(float(value))
````

The first two values are converted and printed. The program then stops at `"n/a"` with a
`ValueError`. Several other types of runtime error have appeared in this course: a `TypeError`
when an operation is applied to a type that does not support it, such as adding a number and
`None`; a `NameError` when a variable that does not exist is used; an `IndexError` when a position
does not exist in a list; a `KeyError` when a key does not exist in a dictionary; and a
`RecursionError` when a function calls itself without end. Each runtime error has a type, and the
type is shown in the error message.

**A logic error.** The cell below is meant to compute the average of four monthly figures.

````python
figures = [4100, 3900, 4400, 3800]

average = sum(figures) / 3
print("Average:", average)
````

The program runs and prints an average of `5400.0`, which is higher than every one of the four
figures. The total was divided by 3 instead of by the number of figures, which is 4. Python
reports nothing, because dividing by 3 is a perfectly valid operation. Logic errors can only be
found by checking whether the results make sense.

The rest of this notebook deals with runtime errors: first how a program can handle them and
continue, then how a function can raise one deliberately.

### Handling errors with try and except

By default, a runtime error stops the program. Data from external sources often contains a few
values that cannot be converted, such as `"n/a"` or an empty text. One way to deal with them is to
check every value before converting it. Another way is to attempt the conversion and handle the
error if it occurs. This is what `try` and `except` do:

- the code inside the `try` block is run;
- if one of its lines raises an error of the type named after `except`, Python stops running the
  `try` block and runs the `except` block instead;
- the program then continues with the line after the whole construction.

The loop below adds up a list of values and counts the ones it cannot convert.

````python
figures = ["38000", "n/a", "39500", "", "47000"]

total = 0
skipped = 0

for value in figures:
    try:
        total = total + float(value)
    except ValueError:
        skipped = skipped + 1

print("Total:", total)
print("Values used:", len(figures) - skipped, "out of", len(figures))
````

The conversion failed for `"n/a"` and for the empty text `""`. Each time, the `except` block
counted the value as skipped, and the loop continued with the next value. Without `try` and
`except`, the program would have stopped at `"n/a"`.

The error type after `except` should always be named. An `except` without a type handles every
kind of error, including errors in the program itself, such as a misspelt variable name. Such a
mistake would then be counted as a skipped value instead of being reported, and it would be hard to
find.

The same construction works for other error types. Looking up a missing key in a dictionary raises
a `KeyError`:

````python
costs = {"salaries": 31000, "rent": 4500}

for key in ["rent", "marketing"]:
    try:
        print(key, costs[key])
    except KeyError:
        print(key, "is not in the cost breakdown")
````

### Raising errors

The runtime errors seen so far were raised by Python itself, when an operation could not be
carried out. A function can also raise an error on purpose. This is useful when the function
receives an argument that Python is able to compute with, but that makes no sense for the task.

The function below applies a discount to an amount. It is called with a discount rate of 2.5.

````python
def apply_discount(amount, rate=0.1):
    discounted = amount * (1 - rate)
    return round(discounted, 2)


print(apply_discount(1000, 2.5))
````

Python raised no error, because the calculation itself is valid: `1000 * (1 - 2.5)` is
`-1500.0`. The problem lies in the meaning of the argument. A discount rate is a share of the
price, so it has to be between 0, for no discount, and 1, for a discount of 100%. A rate of 2.5 is
outside that range, and the result, a negative invoice, is a logic error. Nothing reports it, and a
program using this function would carry on with the wrong figure.

To prevent this, the function can check its argument before using it, and stop with an error if
the argument is not acceptable. The statement that stops the function with an error is `raise`,
followed by the type of error and a message in parentheses:

```python
raise ValueError("rate must be between 0 and 1")
```

`ValueError` is the usual type when an argument has the right type but an unacceptable value. The
message is shown in the error report, so it should say what was wrong.

The version below checks the rate first, and only computes the discount if the rate is
acceptable.

````python
def apply_discount(amount, rate=0.1):
    """Reduce an amount by a rate, which must be between 0 and 1."""
    if rate < 0 or rate > 1:
        raise ValueError("rate must be between 0 and 1")
    discounted = amount * (1 - rate)
    return round(discounted, 2)


print(apply_discount(1000, 0.25))
print(apply_discount(1000, 2.5))
````

The first call has a valid rate, so the check passes and the function returns `750.0`. In the
second call, the check finds a rate above 1, so `raise` stops the function before the calculation
is reached, and Python reports a `ValueError` with the message. The wrong figure that nobody would
have noticed has become an error that is reported at once.

Because it is an ordinary runtime error, it can be handled with `try` and `except`. Writing
`as error` after the error type gives the error a name, and printing that name shows its
message:

````python
try:
    print(apply_discount(1000, 2.5))
except ValueError as error:
    print("Discount not applied:", error)
````

### Handling some errors and passing others on

The previous sections handled errors and raised errors separately. In practice, a function often
does both: it handles the errors it knows how to deal with, and passes on to its caller the errors
it cannot resolve itself.

A `try` block can be followed by several `except` blocks, one for each type of error. When an error
occurs, Python runs the `except` block whose type matches it.

The function below adds up amounts stored as text, all in the same currency, and converts the total
to CHF using a dictionary of exchange rates. The rates are illustrative. Two different errors can
occur inside the loop, and the function treats them differently:

- An amount cannot be read, and `clean_amount` raises a `ValueError`. The function can deal with
  this itself: it skips that amount, reports it, and continues with the others.
- The currency is not in the dictionary, and looking it up raises a `KeyError`. The function cannot
  deal with this: without an exchange rate no total can be computed, and only the caller can
  decide what to do, for example use another rate or stop. The function therefore raises a
  `ValueError` with a clear message, which passes the problem on to the caller.

````python
def total_in_chf(amounts, currency, rates):
    """Add up amounts stored as text, all in one currency, and return the total in CHF.

    Amounts that cannot be read are skipped. A currency without an exchange rate
    is reported to the caller as a ValueError.
    """
    total = 0
    for value in amounts:
        try:
            total = total + clean_amount(value) * rates[currency]
        except ValueError:
            print("Skipped an amount that could not be read:", value)
        except KeyError:
            raise ValueError("no exchange rate for " + currency)
    return round(total, 2)


rates = {"CHF": 1.0, "EUR": 0.94, "USD": 0.80}
amounts = ["1 000", "n/a", "2'500"]

for currency in ["EUR", "GBP"]:
    try:
        print(currency, "total in CHF:", total_in_chf(amounts, currency, rates))
    except ValueError as error:
        print(currency, "total could not be computed:", error)
````

For the amounts in euros, the function skipped `"n/a"` and returned the total of the other two
amounts converted to CHF, `3290.0`. For pounds, there is no exchange rate in the dictionary. The
function raised a `ValueError`, and it was the caller, the loop at the bottom of the cell, that
handled it and printed the message.

Deciding which errors a function handles and which it passes on is part of designing the function.
A function should handle an error itself when it knows the right response. When the right response
depends on the situation of the caller, the function should raise the error and leave the decision
to the caller.
