### How a notebook works

A notebook is a sequence of **cells**. A cell is either **text** (like this one) or **code** (like the one above).

- **Shift + Enter** runs a cell and moves to the next one
- **Ctrl + Enter** runs a cell and stays where you are

The number in brackets to the left of a code cell, `[1]`, `[2]`, tells you the order in which cells were actually run. This matters more than you would think.

### The kernel remembers

Behind the notebook sits a **kernel**: a Python process that stays alive between cells and remembers everything you have defined.

Run the next cell.

```python
capital = 1000
```

Nothing was printed, but the value is now stored. Run the next cell to use it.

```python
print(capital * 1.05)
```

### An experiment worth doing

The kernel remembering is convenient and also the single most common source of confusion.

Do this now:

1. In the menu, choose **Kernel → Restart Kernel**
2. Do **not** re-run the cell that sets `capital`
3. Run the cell below

You should get a `NameError`. That is correct and expected: restarting the kernel wipes its memory, so `capital` no longer exists even though you can still see the cell that created it further up the page.

```python
print(capital * 1.05)
```

### The rule that follows

**What you see on screen is not necessarily what the kernel knows.**

Before you trust a notebook, use **Kernel → Restart Kernel and Run All Cells**. If it works after that, it really works. If it does not, the notebook was relying on something you ran earlier and then changed or deleted.

Do that now, then carry on.
