In this notebook we will look at two main control structures. A program that can **decide**, so that it does
one thing in some circumstances and something else in others. And a program that can **repeat**, so that it
does the same thing many times without you having to write it out many times.

Almost every program you will come across is built from these two ideas.

## Part one — Deciding

### Comparisons

You have already used `+`, `*` and `/`, which take numbers and give you back a number. Comparisons
can also take numbers, but they give you back something different.

````python
print(10 > 3)
print(10 < 3)
print(10 == 10)
````

`True` and `False` are values, in the same way that `7` and `"Neuchâtel"` are values. They
have a type of their own.

````python
print(type(10 > 3))
````

The type is called `bool`, short for boolean, after the mathematician George Boole. There are
only two values of this type.

Here are all the comparisons you will need:

| | meaning |
|---|---|
| `==` | equal to |
| `!=` | not equal to |
| `>` | greater than |
| `<` | less than |
| `>=` | greater than or equal to |
| `<=` | less than or equal to |

Notice that equality is written with two equals signs. A single equals sign means *assign this
value*, which is something quite different. We will come back to it.

````python
revenue = 38000
costs = 52000

print(revenue > costs)
print(revenue != costs)
print(costs - revenue >= 10000)
````

### `if`

A comparison on its own does not do very much. It becomes useful once the program acts on it.

````python
cash = 94000

if cash < 100000:
    print("Cash is below the alert threshold.")
````

You can read that almost as English: *if cash is less than 100,000, then print this.*

Notice the **colon** at the end of the `if` line. If you forget it, Python stops with a
`SyntaxError`.

Notice also the **indentation** on the next line. Those four spaces are what tell Python that the
`print` belongs to the `if`. Python uses indentation where most other languages use brackets, so
it is not just there for readability. It is part of the language.

Now change `cash` to 150000 in the cell above and run it again. You will see that nothing is
printed, and that no error occurs either. The program simply skips over the indented part.

### `if` and `else`

If you also want something to happen in the other case, you add `else`.

````python
cash = 150000

if cash < 100000:
    print("Cash is below the alert threshold.")
else:
    print("Cash is fine.")
````

Exactly one of those two lines will run. You will never get both, and you will never get
neither.

### `elif`

When you have more than two cases, you can use `elif`, which is short for *else if*. You can have
as many as you need.

````python
cash = 94000

if cash < 50000:
    print("Critical")
elif cash < 100000:
    print("Warning")
elif cash < 200000:
    print("Comfortable")
else:
    print("Healthy")
````

Python checks the conditions **in the order you wrote them**, and stops at the first one that
turns out to be true.

That is why the second condition can simply say `cash < 100000`, without also having to say "and
not less than 50,000". If the program has reached that line at all, then the first condition was
already false.

Because of this, the order you choose matters. Try swapping the first two conditions in the cell
above and running it again. You will find that every value below 100,000 now reports "Warning",
and that "Critical" can never be reached at all. The program is still perfectly valid Python and
it still runs quite happily. It is simply wrong.

This is something to keep in mind from now on: a program that runs is not the same thing as a
program that is right.

### Combining conditions

When you need more than one condition at a time, you can join them together. `and` requires both
to be true, `or` requires at least one, and `not` reverses whatever follows it.

````python
revenue = 38000
costs = 52000
cash = 94000

print(revenue < costs and cash < 100000)
print(revenue > costs or cash > 500000)
print(not revenue > costs)
````

````python
if revenue < costs and cash < 100000:
    print("Losing money and running low. Act now.")
````

### A worked example

Let us look at one month of a small company. We have the revenue, the costs, and the cash sitting
in the bank.

The program below decides three things: whether the month was profitable, whether the cash is low,
and which of four situations the business is in.

````python
revenue = 38000
costs = 52000
cash = 94000

profit = revenue - costs

if profit >= 0:
    print("Profitable month:", profit, "CHF")
else:
    print("Loss this month:", -profit, "CHF")

if cash < 100000:
    print("Cash below threshold")

if profit < 0 and cash < 100000:
    print("Both at once. This needs attention.")
elif profit < 0:
    print("Losing money, but there is a buffer.")
elif cash < 100000:
    print("Profitable now, but the buffer is thin.")
else:
    print("Nothing to worry about.")
````

Now change the three numbers at the top and run the cell again, until you have seen all four
of the final messages appear. That is the only way to be sure that every branch actually works,
and it is more or less what testing means.

## Part two — Repeating

### `for`

If you want to do something twelve times, you do not write it out twelve times.

````python
for i in range(5):
    print("month", i)
````

`range(5)` produces 0, 1, 2, 3 and 4. That is five values, starting at zero and stopping
**before** five.

The fact that counting starts at 0 and that the end is excluded will feel arbitrary at first. It
is a convention, it is applied consistently throughout the language, and it is responsible for a
good number of mistakes. You should expect it to catch you out at least once.

You will recognise the colon and the indentation from the `if` section. Here, the indented block
is the part that gets repeated.

````python
for month in range(1, 13):
    print("month", month)
````

If you give `range` two numbers, it starts at the first and stops before the second. So
`range(1, 13)` gives you the twelve months of the year, numbered the way a person would number
them.

### The pattern that matters: accumulating


In the cell bellow, a variable is created **before** the loop, changed **inside** it, and read **after** it.

````python
total = 0

for month in range(1, 13):
    total = total + 4000

print("Total:", total)
````

Follow what happens, slowly.

`total` starts at 0. On each pass through the loop, Python works out the right-hand side using
whatever `total` currently holds, and then stores the result back under the same name. After
twelve passes, `total` holds the sum. The `print` sits outside the loop, so it runs once, at the
end.

There are three things here that are each a separate mistake waiting to happen.

**`total = 0` is outside the loop.** If you move it inside, it resets to zero on every pass, and
what you end up with is one month's figure rather than the total.

**`print` is outside the loop.** If you move it inside, you get twelve lines instead of one.
Sometimes that is exactly what you want, which is why Python cannot warn you either way.

**The order is create, change, read.** Every accumulation you write from now on will have this
same shape, so it is worth getting comfortable with it.

Here is the same pattern used for a running balance rather than a running total. The company
starts with some cash and spends more than it earns.

````python
cash = 180000
revenue = 38000
costs = 52000

for month in range(1, 7):
    cash = cash + revenue - costs
    print(f"After month {month}: {cash} CHF")
````

It is the same shape as before. `cash` is created before the loop and changed inside it. Here
we also print inside the loop, because we want to watch the balance fall month by month.

### A worked example: a loan schedule

A company borrows 120,000 CHF at 4.5% a year and repays it in 24 equal monthly payments.

Each month, part of the payment covers the interest on what is still owed, and the rest reduces
the debt itself. Since the debt gets smaller, the interest gets smaller too, so the split between
the two changes every single month. This is why you cannot answer the question with one formula
and a calculator. You have to walk through it month by month, which is precisely what a loop is
for.

The monthly payment itself comes from the standard annuity formula:

$$\text{payment} = P \times \frac{r}{1 - (1 + r)^{-n}}$$

where $P$ is the amount borrowed, $r$ is the interest rate **per period**, and $n$ is the number
of periods. Here the periods are months, so $r$ is the annual rate divided by twelve and $n$ is
24. The `** -months` in the code below is the $(1 + r)^{-n}$ part of that formula.

````python
loan = 120000
annual_rate = 0.045
months = 24

monthly_rate = annual_rate / 12
payment = loan * monthly_rate / (1 - (1 + monthly_rate) ** -months)

print("Monthly payment:", round(payment, 2), "CHF")
````

````python
balance = loan
total_interest = 0

for month in range(1, months + 1):
    interest = balance * monthly_rate
    principal = payment - interest
    balance = balance - principal
    total_interest = total_interest + interest

    if month <= 3 or month >= 22:
        print("month", month,
              "  interest", round(interest, 2),
              "  principal", round(principal, 2),
              "  still owed", round(balance, 2))

print()
print("Total interest paid:", round(total_interest, 2), "CHF")
````

There are two accumulators running at the same time here, `balance` going down and
`total_interest` going up, and there is an `if` sitting inside the loop so that only the first and
last few months get printed.

That `if` is doing something useful. Without it you would get 24 lines, which is fine for this
loan and quite unreadable for one that runs for 300 months.

Now change `months` to 36 and run both cells again. You will see that the monthly payment falls
and the total interest rises. Nothing else in the program had to change, which is rather the point
of writing it this way.

### `while`

A `for` loop repeats a number of times that you decide in advance. Quite often you do not know
that number, because what you actually want is to keep going *until* something becomes true. That
is what `while` is for.

````python
cash = 180000
month = 0

while cash > 100000:
    month = month + 1
    cash = cash - 14000

print("Cash fell below 100000 CHF in month", month)
````

`while` checks its condition before every pass. As soon as the condition is false, the loop
ends and the program carries on with whatever comes after it.

For a `while` loop to work, you need three things:

1. Something before the loop that sets up the value the condition tests
2. Something **inside** the loop that changes that value
3. The change has to move the value **towards** making the condition false

If any one of the three is missing, the loop will run forever.

### When a loop never stops

This is not a hypothetical problem. It is the most common way to lock up a notebook, and it will
happen to you at some point.

Here is one, written out rather than put in a cell, so that nobody runs it by accident:

```python
cash = 180000
month = 0

while cash > 0:
    month = month + 1          # cash is never changed
```

The condition tests `cash`, but nothing inside the loop ever touches `cash`. So the condition is
true on the first pass, true on the millionth, and true forever after that.

**How you will know it is happening.** The cell shows `[*]` instead of a number and never
finishes. Nothing is broken and you have not lost anything.

**What to do about it.** Press the **stop button** (■) in the toolbar, or use Kernel → Interrupt
Kernel. You will get a `KeyboardInterrupt` message, which is not an error. It is Python confirming
that you stopped it.

If you would like to see it for yourself, type that loop into an empty cell, run it, and then
interrupt it. Doing that once now is much better than meeting it for the first time in the middle
of an exercise.

### A less obvious one

The loop below looks perfectly safe. It changes the value that the condition tests, it moves in
the right direction, and it still never ends.

A company starts with 180,000 CHF and loses money every month, but its revenue grows by 3% a month
while its costs stay flat. We would like to know when the cash runs out.

Read it and decide what you think will happen before you run anything.

```python
cash = 180000
revenue = 38000
costs = 52000
month = 0

while cash > 0:
    month = month + 1
    cash = cash + revenue - costs
    revenue = revenue * 1.03

print("Out of cash in month", month)
```

The cash does fall for a while. Then the revenue overtakes the costs, the company becomes
profitable, and the cash starts climbing again. It never reaches zero, so `cash > 0` never becomes
false, and the loop never ends.

Notice that there is no typo here and nothing is misspelled. The loop is wrong because the question
we asked assumed an outcome that does not actually happen. The program was fine. The assumption
was not.

The way to deal with this is to add a second condition that puts a ceiling on the loop, and then
to check afterwards which of the two ways out it took.

````python
cash = 180000
revenue = 38000
costs = 52000
month = 0

while cash > 0 and month < 36:
    month = month + 1
    cash = cash + revenue - costs
    revenue = revenue * 1.03

if cash <= 0:
    print("Out of cash in month", month)
else:
    print("Still solvent after", month, "months, with", round(cash), "CHF")
    print("Monthly revenue is now", round(revenue), "CHF against costs of", costs, "CHF")
````

As a habit, put a limit on any loop whose stopping condition depends on a calculation. It
costs you one line, and it turns a loop that hangs into an answer you can look at.

The same thing applies to a function that calls itself, which you saw in the first notebook.
Recursion needs a stopping condition for exactly the same reason: something has to become false
eventually, or it never comes back.

### Loops inside loops

You can put a loop inside another loop, and sometimes you will have to. Before you do, it is worth
knowing what it costs you.

The cell below simply counts how many times the innermost line runs.

````python
steps = 0
for month in range(12):
    steps = steps + 1
print("one loop over 12 months:", steps, "steps")

steps = 0
for month in range(12):
    for scenario in range(20):
        steps = steps + 1
print("a loop inside a loop:    ", steps, "steps")

steps = 0
for month in range(12):
    for scenario in range(20):
        for product in range(15):
            steps = steps + 1
print("three deep:              ", steps, "steps")
````

You get 12, then 240, then 3,600.

Notice that each level does not add work, it **multiplies** it. A third loop over a fairly modest
15 items turned a twelve-step program into an over three-thousand-step one.

At these sizes it makes no difference at all, because your computer does millions of simple steps a
second. The reason to pay attention is that the numbers in real work are not 12 and 20.
Twelve months, twenty scenarios and fifteen products is a small question. Ten thousand customers,
three hundred days and fifty products is an ordinary one, and that comes to 150 million steps.

So here is a habit to form now. Before you put a loop inside a loop, ask yourself
whether you really need every combination of the two things, or whether you only need to do two
separate jobs.


````python
# two jobs, one after the other: 12 + 12 steps
monthly_revenue = 38000
monthly_costs = 52000

total_revenue = 0
for month in range(12):
    total_revenue = total_revenue + monthly_revenue

total_costs = 0
for month in range(12):
    total_costs = total_costs + monthly_costs

print("Revenue", total_revenue, "costs", total_costs, "result", total_revenue - total_costs)
````

There is an entire field devoted to reasoning about this properly, and it is not part of this
course. The rule of thumb to take away is that one loop is cheap, a loop inside a loop is a good
deal more expensive than it looks, and three levels deep is usually a sign that there is a better
way to ask the question.

**9.** This last one puts everything together.

The program below follows a company through one year. Revenue starts at 38,000 CHF a month and
grows by 6% each month, costs stay flat at 52,000, and the company starts with 180,000 CHF in the
bank. It is supposed to report three things: how many months were loss-making, the lowest the cash
got at any point during the year, and the cash left at the end.

It runs without any error. Two of the three numbers it prints are wrong.

There are two separate mistakes in it. One of them is an ordinary logical error. The other is a
line sitting at the wrong indentation level, so it belongs to the wrong block and only runs some
of the time.

Find both, fix them, and for each one write a sentence saying what it did to the answer.

````python
cash = 180000
revenue = 38000
costs = 52000
growth = 1.06

lowest_cash = 0
months_in_the_red = 0

for month in range(1, 13):
    profit = revenue - costs

    if profit < 0:
        months_in_the_red = months_in_the_red + 1
        cash = cash + profit

    if cash < lowest_cash:
        lowest_cash = cash

    revenue = revenue * growth

print("Months in the red:", months_in_the_red)
print("Lowest cash during the year:", round(lowest_cash))
print("Cash at the end:", round(cash))
````
