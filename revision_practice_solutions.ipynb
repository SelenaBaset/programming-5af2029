## Exercise 1: Two weeks of weather in Neuchâtel

**The data.** The list `temperatures` below holds the temperature measured at noon in Neuchâtel on

**The questions.** Spring weather changes quickly: a few warm days can be followed by a night of

```python
temperatures = [6.5, 8.0, 11.5, 12.0, 9.5, 3.0, -1.5, -0.5, 4.0, 10.5, 12.5, 13.0, 11.0, 7.5]
```

**a. Read: the average temperature**

**b. Read: the warmest day**

The cell below finds the warmest day.

```python
warmest = 0
for day in range(len(temperatures)):
    if temperatures[day] > temperatures[warmest]:
        warmest = day

print("Warmest day: day", warmest + 1, "with", temperatures[warmest], "degrees")
```

```python
# b. Answers
#
# 1. warmest holds the position in the list of the warmest day found so far.
#
# 2. day and warmest are positions, not temperatures. Comparing them would only tell which day
#    comes later in the list. temperatures[day] and temperatures[warmest] are the temperatures
#    at these positions, which is what has to be compared.
#
# 3. The positions in a list start at 0, but the days are numbered from 1.
#    The warmest day is at position 11, which is day 12.
```

**c. Write: the coldest day and the frost days**

```python
# c. The coldest day and the frost days

# 1. The same code as for the warmest day, with < instead of >.
coldest = 0
for day in range(len(temperatures)):
    if temperatures[day] < temperatures[coldest]:
        coldest = day

print("Coldest day: day", coldest + 1, "with", temperatures[coldest], "degrees")

# 2. The same shape as the average, but the loop counts instead of adding temperatures,
#    and it only counts the days that pass the test.
frost_days = 0
for temperature in temperatures:
    if temperature < 0:
        frost_days = frost_days + 1

print("Frost days:", frost_days)
```

**d. Read, then write: the longest warm run**

```python
# d. The longest run below 4 degrees

# The corrected code of d, with the test changed from "above 10" to "below 4".
current_run = 0
longest_run = 0

for temperature in temperatures:
    if temperature < 4:
        # One more cold day in a row.
        current_run = current_run + 1
        if current_run > longest_run:
            longest_run = current_run
    else:
        # The run is broken: start counting again from 0.
        current_run = 0

print("Longest run below 4 degrees:", longest_run, "days")
```

## Exercise 2: Is this password strong enough?

<img src="https://images.unsplash.com/photo-1774898989484-0b9becf69efb?fm=jpg&q=70&w=800&auto=format&fit=crop" alt="A padlock on a computer keyboard" align="right" width="340" hspace="16">

**The rules.** A website checks every new password against three rules:

- it has at least 10 characters;
- it contains at least one digit;
- it contains at least one uppercase letter.

**The verdict.** A password that meets none or one of the rules is *weak*, one that meets two is
*medium*, and one that meets all three is *strong*.

<sub>Picture: Sasun Bughdaryan on <a href="https://unsplash.com/photos/a-padlock-rests-on-a-computer-keyboard-QP7RBa5r8HM">Unsplash</a>, Unsplash License.</sub>

```python
passwords = ["sunshine", "Sunshine2026", "neuchatel", "lake-view-77",
             "CorrectHorseBattery", "A1b2", "Lake-View-77!"]
```

**a. Read: checking the first two rules**

The cell below defines two functions. The second uses the string method `isdigit`, which is `True`
when a character is a digit: `"7".isdigit()` gives `True`, and `"a".isdigit()` gives `False`.

1. What does `has_digit("abc1def")` return? How many characters does the loop look at before the
   function returns?
2. What would `has_digit("abc1def")` return if the line `return False` were indented by four more
   spaces, so that it is inside the loop? Why?

```python
def long_enough(password):
    """Return True if the password has at least 10 characters."""
    if len(password) >= 10:
        return True
    else:
        return False


def has_digit(password):
    """Return True if the password contains at least one digit."""
    for character in password:
        if character.isdigit():
            return True
    return False


print(long_enough("sunshine"), long_enough("Sunshine2026"))
print(has_digit("sunshine"), has_digit("abc1def"))
```

```python
# a. Answers
#
# 1. It returns True. The loop looks at "a", "b", "c" and "1".
#    At "1", return True ends the function at once, so "d", "e" and "f" are never looked at.
#
# 2. It would return False. With return False inside the loop, the function ends at the first
#    character that is not a digit. "a" is not a digit, so the function returns False before
#    it reaches "1". return False must come after the loop, once every character has been checked.
```

**b. Write: the third rule**

Write a function `has_uppercase(password)` that returns `True` if the password contains at least
one uppercase letter. Follow `has_digit`, and use the string method `isupper`:
`"B".isupper()` gives `True`, and `"b".isupper()` gives `False`.

```python
# b. The function has_uppercase

# The same code as has_digit, with isupper() instead of isdigit().
def has_uppercase(password):
    """Return True if the password contains at least one uppercase letter."""
    for character in password:
        if character.isupper():
            # One uppercase letter is enough.
            return True

    # The loop has looked at every character and found no uppercase letter.
    return False


print(has_uppercase("neuchatel"), has_uppercase("Sunshine2026"))
```

**c. Read: the verdict**

The cell below combines the three rules.

1. Without running it, predict the verdict for `"A1b2"`, `"CorrectHorseBattery"` and `"neuchatel"`.
2. Run it and check your predictions. Is any verdict surprising?

```python
def strength(password):
    """Return "weak", "medium" or "strong", depending on how many rules the password meets."""
    rules_met = 0
    if long_enough(password):
        rules_met = rules_met + 1
    if has_digit(password):
        rules_met = rules_met + 1
    if has_uppercase(password):
        rules_met = rules_met + 1

    if rules_met == 3:
        return "strong"
    elif rules_met == 2:
        return "medium"
    else:
        return "weak"


for password in passwords:
    print(password, "->", strength(password))
```

```python
# c. Answers
#
# 1. "A1b2": a digit and an uppercase letter, but only 4 characters. Two rules: medium.
#    "CorrectHorseBattery": long and with uppercase letters, but no digit. Two rules: medium.
#    "neuchatel": short, no digit, no uppercase letter. No rule: weak.
#
# 2. "A1b2" is surprising: a 4-character password gets the same verdict as a 19-character one.
#    Each rule counts the same, so the verdict does not measure how hard a password is to guess.
```

**d. Write: a new rule**

The website adds a fourth rule: the password must contain at least one of the symbols in
`symbols = "!?-_@#"`. The verdict changes too: a password is *strong* if it meets all four rules,
*medium* if it meets two or three, and *weak* otherwise.

1. Write a function `has_symbol(password)`. The test `character in symbols` is `True` if the
   character is one of the symbols.
2. Write a new version of `strength` with the four rules and the new verdicts, and print the verdict
   for each password of the list.

> **Check:** `"Lake-View-77!"` is the only strong password.

```python
# d. A fourth rule

symbols = "!?-_@#"


def has_symbol(password):
    """Return True if the password contains at least one of the symbols."""
    for character in password:
        # in checks whether the character appears in the text symbols.
        if character in symbols:
            return True
    return False


def strength(password):
    """Return "weak", "medium" or "strong", depending on how many of the four rules are met."""
    rules_met = 0
    if long_enough(password):
        rules_met = rules_met + 1
    if has_digit(password):
        rules_met = rules_met + 1
    if has_uppercase(password):
        rules_met = rules_met + 1
    if has_symbol(password):
        rules_met = rules_met + 1

    # The new verdicts: 4 rules for strong, 2 or 3 for medium.
    if rules_met == 4:
        return "strong"
    elif rules_met >= 2:
        return "medium"
    else:
        return "weak"


for password in passwords:
    print(password, "->", strength(password))
```

**a. Read: from text to seconds**

The cell below turns a duration into a number of seconds. It uses the string method `split`, which
cuts a text into a list of pieces wherever it finds the separator given in brackets.

1. For the duration `"3:45"`, what does the list `parts` contain? Is `parts[0]` a number or a text?
2. Why are `int(parts[0])` and `int(parts[1])` needed?
3. Remove the `#` in front of the last line, and run the cell again. Which error does Python raise
   for `"4:7x"`, and on which line?

```python
def to_seconds(duration):
    """Turn a duration written as "minutes:seconds" into a number of seconds."""
    parts = duration.split(":")
    minutes = int(parts[0])
    seconds = int(parts[1])
    return minutes * 60 + seconds


print(to_seconds("3:45"))
# print(to_seconds("4:7x"))
```

```python
# a. Answers
#
# 1. parts is the list ["3", "45"]. parts[0] is the text "3", not the number 3.
#
# 2. Texts cannot be used in a calculation: "3" * 60 would repeat the text 60 times.
#    int() turns each text into a whole number, so that minutes * 60 + seconds can be computed.
#
# 3. Python raises a ValueError, on the line seconds = int(parts[1]).
#    The text "7x" is not a whole number, so int() cannot convert it.
```

**b. Write: from seconds to text**

Write a function `to_text(seconds)` that does the opposite of `to_seconds`: it turns a number of
seconds into a text such as `"3 min 45 s"`.

The operator `//` gives the result of a division without the decimals, and `%` gives what is left
over: `225 // 60` gives 3, the number of whole minutes, and `225 % 60` gives 45, the seconds left.

> **Check:** `to_text(225)` gives `"3 min 45 s"`.

```python
# b. The function to_text

def to_text(seconds):
    """Turn a number of seconds into a text such as "3 min 45 s"."""
    # // gives the number of whole minutes, and % gives the seconds left over.
    minutes = seconds // 60
    rest = seconds % 60

    # str() turns a number into text, so that it can be joined to other texts with +.
    return str(minutes) + " min " + str(rest) + " s"


print(to_text(225))
```

**d. Read: the songs played during the trip**

The cell below chooses the songs played during the 30-minute trip.

1. Run it. Which songs are played, and how much time is left at the end?
2. What does the `continue` inside `except` do here?
3. Replace `break` by `continue`, and run the cell again. Which song is added, and why? Which of the
   two versions follows the rule of the trip, "a song is only started if it can finish before the
   train arrives", while keeping the order of the playlist?

```python
trip = 30 * 60      # the length of the trip, in seconds
used = 0            # the time already filled with songs, in seconds

for song in playlist:
    try:
        seconds = to_seconds(playlist[song])
    except ValueError:
        continue

    if used + seconds > trip:
        break

    used = used + seconds
    print(song, "(" + playlist[song] + ")")

print()
print("Time left:", to_text(trip - used))
```

```python
# d. Answers
#
# 1. Six songs are played, from "Morning on the Lake" to "Chocolate and Coffee",
#    and 6 min 41 s are left.
#
# 2. continue skips the rest of the loop for this song and goes straight to the next song.
#    The two songs whose duration cannot be read are therefore skipped.
#
# 3. With continue, "Bern by Night" is added, and 1 min 57 s are left.
#    "Through the Tunnel" (7:20) does not fit. break stops the loop there, while continue only
#    skips this song and goes on: "Bern by Night" (4:44) is shorter and still fits.
#
#    Both versions only start a song that can finish in time. Only the version with break also
#    keeps the order of the playlist: it stops at the first song that does not fit. The version
#    with continue skips that song and fills the time with a later one.
```

**d. Read: counting all the pairs**

Two dice can land in 36 different ways: (1, 1), (1, 2), and so on up to (6, 6). The cell below goes
through all of them with one loop inside another.

1. How many times does the line `pairs[total] = pairs[total] + 1` run in total? Why?

```python
pairs = {}
for total in range(2, 13):
    pairs[total] = 0

for first in range(1, 7):
    for second in range(1, 7):
        total = first + second
        pairs[total] = pairs[total] + 1

for total in range(2, 13):
    exact = pairs[total] / 36 * 100
    print("Sum", total, ":", pairs[total], "of 36 pairs, exact chance", round(exact, 2), "%")
```

```python
# d. Answers
#
# 1. 36 times. The outer loop runs 6 times, once for each value of the first die,
#    and each time the inner loop runs 6 times, once for each value of the second die: 6 x 6 = 36.
#
# 2. (1, 3), (2, 2) and (3, 1). pairs[4] is 3.
#
# 3. The shares from the 10,000 rolls are close to the exact chances, but not equal to them.
#    For example, the exact chance of 7 is 16.67%, and the rolls give a share close to it.
#    The more rolls, the closer the shares usually get.
```

## Exercise 5: Secret messages

<img src="https://images.unsplash.com/photo-1762930518943-d931fc04a019?fm=jpg&q=70&w=800&auto=format&fit=crop" alt="The keyboard of an Enigma machine" align="right" width="340" hspace="16">

**The Caesar cipher.** One of the oldest ways to hide a message is to shift every letter by the same
number of places in the alphabet. With a shift of 3, A becomes D, B becomes E, and so on. At the end
of the alphabet, the count starts again from A: X becomes A, Y becomes B and Z becomes C.

**Decrypting.** To read the message, the receiver shifts every letter back by the same number of
places. Spaces and punctuation are left as they are.

**The messages.** All the messages in this exercise are written in capital letters.

<sub>Picture: Markus Kammermann, the keyboard of an Enigma cipher machine, on <a href="https://unsplash.com/photos/close-up-of-an-old-enigma-machine-keyboard-YHICQTbKSzc">Unsplash</a>, Unsplash License.</sub>

```python
secret = "XPPE XP LE ESP ZWO WTMCLCJ LE YZZY LYO MCTYR ESP YZEPD"

common_words = ["THE", "AND", "AT", "ME", "TO", "IS", "OF", "IN", "IT", "A", "YOU", "ON", "FOR"]
```

**a. Read: shifting one letter**

The cell below shifts one letter. It uses the string method `index`, which gives the position of a
character in a text: `alphabet.index("D")` gives 3, because A is at position 0.

1. Work out by hand what `shift_letter("Y", 3)` returns. Write down the value of `position` and
   of `new_position`.
2. Do the same for `shift_letter("B", -3)`. In Python, `(1 - 3) % 26` gives 24.
3. What is `% 26` for?

Then run the cell to check your answers.

```python
alphabet = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"


def shift_letter(letter, shift):
    """Return the letter shifted by a number of places in the alphabet."""
    position = alphabet.index(letter)
    new_position = (position + shift) % 26
    return alphabet[new_position]


print(shift_letter("Y", 3), shift_letter("B", -3))
```

```python
# a. Answers
#
# 1. Y is at position 24. new_position is (24 + 3) % 26, that is 27 % 26, which is 1.
#    alphabet[1] is "B".
#
# 2. B is at position 1. new_position is (1 - 3) % 26, which is 24. alphabet[24] is "Y".
#
# 3. % 26 keeps the new position between 0 and 25. A position past Z (27) starts again from A,
#    and a position before A (-2) starts again from Z. Without it, alphabet[27] would raise an
#    IndexError.
```

3. In the empty cell, write a function `decrypt(message, shift)` that undoes the encryption. It
   needs only one line, which calls `encrypt`.

> **Check:** `decrypt("KHOOR, ZRUOG!", 3)` gives `"HELLO, WORLD!"`.

```python
def encrypt(message, shift):
    """Shift every letter of the message, and leave the other characters unchanged."""
    result = ""
    for character in message:
        if character in alphabet:
            result = result + shift_letter(character, shift)
        else:
            result = result + character
    return result


print(encrypt("HELLO, WORLD!", 3))
```

```python
# b. Answers
#
# 1. They are kept as they are. The test if character in alphabet is False for them,
#    so the else adds the character to the result without shifting it.
#
# 2. The result is built one character at a time, by adding each character to the end of it.
#    It has to start as an empty text, so that the first character can be added to something.
```

```python
# b. The function decrypt

def decrypt(message, shift):
    """Undo the shift of an encrypted message."""
    # Decrypting is encrypting with the opposite shift: shifting back by 3 is shifting by -3.
    return encrypt(message, -shift)


print(decrypt("KHOOR, ZRUOG!", 3))
```

**d. Read, then write: letting the program choose**

Reading 26 attempts works for one message, but not for a thousand. The function below gives the
program a way to recognise English: it counts how many words of a text are in the list
`common_words`. The string method `split(" ")` cuts a text into a list of words at each space.

1. What does `count_common("MEET ME AT THE OLD LIBRARY")` return? Which words are counted?
2. In the empty cell, write a loop that tries every shift, and keeps the attempt with the highest
   count, together with its shift. This is the same best-so-far method as for the warmest day and
   the most frequent sum. Print the shift and the message.

> **Check:** the right attempt contains 6 words from `common_words`.

```python
def count_common(text):
    """Return how many words of the text are in the list common_words."""
    count = 0
    for word in text.split(" "):
        if word in common_words:
            count = count + 1
    return count


print(count_common("MEET ME AT THE OLD LIBRARY"))
```

```python
# d. Answers
#
# 1. It returns 3: "ME", "AT" and "THE" are in common_words.
#    "MEET", "OLD" and "LIBRARY" are not.
```

```python
# d. Finding the right shift automatically

# The best attempt found so far, its shift, and its number of common words.
best_attempt = ""
best_shift = 0
best_count = 0

for shift in range(26):
    attempt = decrypt(secret, shift)
    count = count_common(attempt)

    # Keep this attempt if it has more common words than the best one so far.
    if count > best_count:
        best_attempt = attempt
        best_shift = shift
        best_count = count

print("Shift:", best_shift)
print("Message:", best_attempt)
print("Common words found:", best_count)
```
