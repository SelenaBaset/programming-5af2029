## Where you will get the course material updates

All the course material lives in a [public repository on GitHub](https://github.com/SelenaBaset/programming-5af2029). New notebooks are added to your own workspace when you click the course link on Moodle again. Your edits and your answers are kept: nothing you have written is overwritten. Bookmark that link so you do not have to go through Moodle each time.

Alternatively, you can also update from inside Noto: open the Git panel in the left sidebar and click Pull, or use git pull from a terminal. Both work, but the course link handles the awkward case better. When you have edited a notebook that I have also changed, the link keeps your version and merges mine in, while git pull will stop and ask you to resolve the conflict yourself.

## Where to run it

The supported environment is **Noto** (`noto.epfl.ch`), accessed with your UniNE account. Everything this course uses is already installed there.

Google Colab or a local installation also work, but are not supported. If you use one, you look after your own setup.

## Assessment

One **written exam, on paper**, 120 minutes. Open book: all course documentation, including your own notes. No personal electronic device, no internet, no AI assistant, no communication with anyone else.

What is tested: reading code, predicting results, spotting and fixing faults, and adapting a program to a changed requirement.

Work during the semester is **not graded**. It is how you get ready.

## Using AI

You are encouraged to use AI assistants in this course. Used well, they let you cover far more ground than the syllabus alone: ask why a line works, ask for a second example, ask what would happen if you changed something, ask about a topic we did not have time for.

The one thing worth avoiding is taking an answer you cannot follow. If you cannot explain what the code does, you have the answer without the understanding, and the understanding is the part that transfers. The exam is on paper with no devices, which is simply a reflection of that.
