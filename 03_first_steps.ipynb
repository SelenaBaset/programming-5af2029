### Types

Every value has a **type**. The three you meet first:

```python
print(type(10000))        # int, a whole number
print(type(0.03))         # float, a decimal number
print(type("Neuchâtel"))  # str, text
```

Types decide what operations mean. The same `+` does two different things here.

```python
print(3 + 4)
print("3" + "4")
```

And sometimes it refuses. Run the next cell: it is **meant** to fail.

Read the last line of the error before moving on.

```python
print("3" + 4)
```

`TypeError: can only concatenate str (not "int") to str`

Python is telling you precisely what went wrong: you asked it to join text to a number, and it does not guess what you meant. Getting used to reading these messages is most of what learning to debug means.

You can convert between types explicitly.

```python
print("3" + str(4))
print(int("3") + 4)
```

### Putting values into text

An **f-string** is text with values slotted into it. Put an `f` before the quote and the expression in curly braces.

```python
capital = 10000
rate = 0.03

print(f"An investment of {capital} CHF at {rate:.1%} earns {capital * rate:.2f} CHF in a year.")
```

`:.2f` means two decimal places, `:.1%` formats as a percentage. You will use this constantly.

### A small calculation

Compound interest, written out step by step.

```python
capital = 10000
rate = 0.03
years = 5

final = capital * (1 + rate) ** years

print(f"After {years} years: {final:.2f} CHF")
print(f"Total gain:        {final - capital:.2f} CHF")
```
